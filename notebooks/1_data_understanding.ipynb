# Exploratory Data Analysis for Linear Regression

## Objective
The objective of this notebook is to understand the structure and behavior of the Advertising dataset before applying Linear Regression models. This analysis includes examining dataset properties, checking for missing and duplicate values, understanding feature distributions, visualizing relationships between variables, and identifying initial patterns that may influence regression modeling.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# load dataset
data = pd.read_csv('/content/drive/My Drive/Advertising.csv')
data
```

```python
# dataset overview
data.info()
```

```python
# checking duplicate values
data.duplicated().sum()
```

```python
# statistical overiew of dataset
data.describe()
```

```python
# distribution analysis
fig, ax = plt.subplots(1, 3, figsize=(17, 7))

sns.histplot(data['TV'], kde=True, color='r', ax=ax[0])
sns.histplot(data['Newspaper'], kde=True, color='g', ax=ax[1])
sns.histplot(data['Radio'], kde=True, color='b', ax=ax[2])

plt.show()
```

```python
# linearity analysis
fig, ax = plt.subplots(1, 3, figsize=(17, 7))

ax[0].scatter(data['TV'], data['Sales'], color='r')
ax[0].set_xlabel('TV')
ax[0].set_ylabel('Sales')

ax[1].scatter(data['Newspaper'], data['Sales'], color='g')
ax[1].set_xlabel('Newspaper')
ax[1].set_ylabel('Sales')

ax[2].scatter(data['Radio'], data['Sales'], color='b')
ax[2].set_xlabel('Radio')
ax[2].set_ylabel('Sales')

plt.show()
```

```python
# correlation analysis
data[['TV', 'Radio', 'Newspaper']].corr()
```

## Key Observations


1. The dataset contains 200 rows and 4 columns. The independent variables are TV, Radio, and Newspaper advertising budgets, while Sales is the dependent variable. All columns contain numerical (float) data types, and no missing values are present in the dataset.

2. No duplicate rows were found in the dataset, indicating that each observation is unique.

3. The average spending on TV advertising is the highest among all platforms, followed by Newspaper and Radio advertising. TV advertising also has the highest standard deviation, which indicates that TV advertising budgets vary significantly across observations. The Newspaper feature appears positively skewed, meaning most observations contain lower newspaper advertising budgets, while a few observations show very high spending values. The Radio feature appears more evenly distributed within its range.

4. Scatter plots between individual advertising platforms and Sales show that TV and Radio advertising have a relatively stronger positive linear relationship with Sales. In contrast, the Newspaper feature shows a weaker and less clear linear relationship with the target variable.

5. Based on the scatter plots and distribution analysis, the Newspaper feature may have lower predictive importance in a Linear Regression model because its relationship with Sales appears weaker compared to TV and Radio advertising.

6. The correlation table suggests that independent variables are not strongly correlated with one another, indicating that multicollinearity is not present in the dataset. Therefore, the dataset appears suitable for applying a baseline Linear Regression model.
